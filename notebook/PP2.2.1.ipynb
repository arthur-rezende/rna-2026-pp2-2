3.2 Escolha uma das métricas apresentadas para ser utilizada como referência pela equipe

```python
atributos = X.columns

for col in atributos:
    fig, (ax_box, ax_hist) = plt.subplots(
        2, 1, figsize=(8, 4), sharex=True,
        gridspec_kw={"height_ratios": [1, 3]}
    )

    # boxplot: mediana, quartis, dispersão e outliers
    ax_box.boxplot(X[col], vert=False, widths=0.6)
    ax_box.set_yticks([])
    ax_box.set_title(col)

    # histograma: forma da distribuição
    ax_hist.hist(X[col], bins=40, edgecolor="white")
    ax_hist.axvline(X[col].mean(), color="red", linestyle="--", label=f"média = {X[col].mean():.3f}")
    ax_hist.axvline(X[col].median(), color="green", linestyle="-", label=f"mediana = {X[col].median():.3f}")
    ax_hist.legend()
    ax_hist.set_ylabel("Frequência")

    plt.tight_layout()
    plt.show()
```

1.2 Medidas estatísticas

```python
estatisticas = X.describe().T
estatisticas["IQR"] = estatisticas["75%"] - estatisticas["25%"]
estatisticas["assimetria"] = X.skew()
display(estatisticas[["mean", "50%", "std", "min", "max", "IQR", "assimetria"]].round(3))
```

### 2. Quanto ao balanceamento do atributo alvo

```python
contagem = df["Class"].value_counts()
proporcao = df["Class"].value_counts(normalize=True) * 100

ax = contagem.plot(kind="bar", figsize=(8, 4), edgecolor="black")
ax.set_title("Distribuição das classes (Class)")
ax.set_ylabel("Número de exemplos")
for i, (n, p) in enumerate(zip(contagem, proporcao)):
    ax.text(i, n, f"{n}\n({p:.1f}%)", ha="center", va="bottom", fontsize=8)
plt.tight_layout()
plt.show()

print("Razão maior/menor classe:", round(contagem.max() / contagem.min(), 2))
```

R: A base é desbalanceada, a variante de feijão DERMASON é a maior classe com cerca de 26% dos exemplos e a BOMBBAY é a menor com menos de 4%. Logo o modelo pode ter uma alta acurácia só acertando as classes grades e errando as pequenas.
